# Exploratory Data Analysis - Car Price Prediction

This notebook contains exploratory analysis on `cardekho_dataset.csv` including missing value analysis, distribution plots, correlation analysis, and feature selection.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Load dataset
df = pd.read_csv('../Dataset/cardekho_dataset.csv')
df.head()
```

```python
print("Dataset Info:")
df.info()
print("\nSummary Statistics:")
df.describe()
```

```python
print("Missing values check:")
print(df.isnull().sum())
print("\nDuplicated rows:", df.duplicated().sum())
```

```python
# Categorical and Numerical features breakdown
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
print("Numerical Columns:", num_cols)
print("Categorical Columns:", cat_cols)
```

```python
# Visualizing selling price distribution
plt.figure(figsize=(10, 5))
sns.histplot(df['selling_price'], kde=True, bins=50, color='purple')
plt.title('Distribution of Selling Price')
plt.xlabel('Selling Price (INR)')
plt.show()
```
